```python
# Selection Implementation
import random

def random_partition(arr, low, high):
    """
    Selects a random pivot, partitions the array around it,
    and places all elements <= pivot on the left and > pivot on the right.
    """
    # Pick random pivot and swap with high
    pivot_idx = random.randint(low, high)
    arr[pivot_idx], arr[high] = arr[high], arr[pivot_idx]
    pivot = arr[high]
    
    i = low - 1
    for m in range(low, high):
        if arr[m] <= pivot:
            i += 1
            arr[i], arr[m] = arr[m], arr[i]
            
    # Place pivot in its sorted boundary position
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1


def select(arr, low, high, k):
    """
    Finds the k-th smallest element (1-based rank) in arr[low ... high].
    Operates in expected O(n) time.
    """
    if low == high:
        return arr[low]
    
    p = random_partition(arr, low, high)
    
    # j is the number of elements in arr[low ... p] (including pivot)
    j = (p - low) + 1
    
    if k == j:
        return arr[p]
    elif k < j:
        # Search left of pivot for k-th smallest
        return select(arr, low, p - 1, k)
    else:
        # Search right of pivot and account for left of pivot being discarded (k-j)
        return select(arr, p + 1, high, k - j)

```

```python
# Tests
def check_partition(arr):
    p = random_partition(arr, 0, len(arr) - 1)
    pivot = arr[p]
    assert all(x <= pivot for x in arr[:p])
    assert all(x > pivot for x in arr[p + 1:])

def check_select(arr):
    for k in range(1, len(arr) + 1):
        assert select(list(arr), 0, len(arr) - 1, k) == sorted(arr)[k - 1]

cases = [
    [5],                          # single element
    [2, 1],                       # two elements
    [1, 2, 3, 4, 5],              # already sorted
    [5, 4, 3, 2, 1],              # reverse sorted
    [4, 4, 4, 4],                 # all equal
    [3, 1, 3, 2, 3, 1],           # duplicates
    [-3, 5, -1, 0, -7, 2],        # negatives
    [9, 4, 7, 1, 8, 2, 6, 3, 5],  # typical
]

for case in cases:
    for _ in range(20):           # pivot is random, so repeat each case
        check_partition(list(case))
    check_select(case)

print("All tests passed.")
```
